```python
payload = {"username": "ass", "password": "we can"}
r = requests.post("https://pythonscraping.com/pages/cookies/welcom.php", data=payload)
print(r.cookies.get_dict())
r = requests.get("https://pythonscraping.com/pages/cookies/profile.php", r.cookies)
print(r.text)
```

```python
session = requests.Session()
payload = {"username": "ass", "password": "we can"}
r = session.post("https://pythonscraping.com/pages/cookies/welcom.php", data=payload)
print(r.cookies.get_dict())
r = session.get("https://pythonscraping.com/pages/cookies/profile.php")
print(r.text)
```
