```python
from urllib.request import urlopen

from bs4 import BeautifulSoup

html = (
    urlopen("https://MorvanZhou.github.io/static/scraping/basic-structure.html")
    .read()
    .decode("utf-8")
)
print(html)
```

```python
soup = BeautifulSoup(html, features="lxml")
print(soup.h1)
print("\n", soup.p)
```

```python
all_href = soup.find_all("a")
all_href = [l["href"] for l in all_href]
print(all_href)
```
