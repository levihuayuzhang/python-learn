```python
plt.figure(figsize=(10, 5))

month = ["jan", "feb", "mar", "apr"]
sales = [100, 150, 80, 130]

plt.plot(
    month,
    sales,
    label="product A",
    color="orange",
    linewidth="2",
    linestyle="--",
    marker="o",
)
plt.title("2025 sales trend", color="red", fontsize=20)
plt.xlabel("month", fontsize=10)
plt.ylabel("sale (10 thouthands yuan)", fontsize=10)
plt.legend(loc="upper left")
plt.grid(True, alpha=0.3, color="blue", linestyle="--")  # plt.grid(axis="y")
plt.xticks(rotation=60, fontsize=12)
plt.yticks(rotation=0, fontsize=12)
plt.ylim(0, 160)
for x, y in zip(month, sales):
    print(x, y)
    plt.text(x, y + 1, str(y), ha="center", va="bottom", fontsize=10)

plt.show()
```

```python
subjects = ["chinese", "math", "english", "science"]
scores = [85, 92, 78, 88]

# plt.barh()
plt.bar(subjects, scores, label="score", color="purple", width=0.6)
plt.title("Lil Red's score distribution in 2025", color="green", fontsize=15)
plt.xlabel("subjects", fontsize=10)
plt.ylabel("scores", fontsize=10)
plt.legend(loc="upper right")
plt.grid(axis="y", alpha=0.3, color="blue", linestyle="--")
plt.xticks(rotation=0, fontsize=12)
plt.yticks(rotation=0, fontsize=12)
plt.ylim(0, 100)
for x, y in zip(subjects, scores):
    print(x, y)
    plt.text(x, y + 1, str(y), ha="center", va="bottom", fontsize=10)
plt.tight_layout()

plt.show()
```

```python
things = ["learn", "entertainment", "sport", "sleep", "others"]
times = [6, 4, 1, 8, 5]

colors = ["#66b3ff", "#99ff99", "#ffcc99", "#ff9999", "#ff4499"]
explode = [0.1, 0, 0, 0, 0]
plt.pie(
    times,
    labels=things,
    autopct="%.2f%%",
    startangle=90,
    colors=colors,
    wedgeprops={"width": 0.6},
    pctdistance=0.8,
    explode=explode,
    shadow=True,
)
plt.title("time distribution for a day")
plt.text(0, 0, "total:\n100%", ha="center", va="center", fontsize=10)

plt.tight_layout()

plt.show()
```

```python
# scores = [50, 55, 60, 65, 70, 75, 80]
# hours = [1, 2, 3, 4, 5, 6, 7]
# plt.scatter(hours, scores)

import random

x = []
y = []
for i in range(300):
    tmp = random.uniform(0, 10)
    x.append(tmp)
    y.append(2 * tmp + random.gauss(0, 2))

plt.scatter(x, y, color="blue", alpha=0.5, s=20, label="data")
plt.title("relationship of x and y")
plt.xlabel("x")
plt.ylabel("y")
plt.legend(loc="upper left")
plt.grid(True, alpha=0.3, linestyle="--")
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.ylim(0, 30)

plt.plot([0, 10], [0, 20], color="red", linewidth=2, linestyle="-")

plt.tight_layout()

plt.show()
```

```python
data = {
    "chinese": [82, 85, 88, 70, 90, 76, 84, 83, 95],
    "math": [75, 80, 79, 93, 88, 82, 87, 89, 92],
    "english": [70, 72, 68, 65, 78, 80, 85, 90, 95],
}

plt.figure(figsize=(8, 6))
plt.boxplot(data.values(), tick_labels=data.keys())

plt.title("boxplot for subjects scores distribution")
plt.ylabel("scores")
plt.grid(True, axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```
