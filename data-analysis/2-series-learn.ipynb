```python
import pandas as pd

s = pd.Series([1, 2, 3, 4, 5])
print(s)

s = pd.Series([10, 2, 3, 4, 5], index=["A", "B", "C", "D", "E"])
print(s)

s = pd.Series([10, 2, 3, 4, 5], index=["A", "B", "C", "D", "E"], name="month")
print(s)
```

```python
s = pd.Series({"A": 1, "B": 2, "C": 3, "D": 4, "E": 5})
print(s)

s1 = pd.Series(s, index=["A", "C"])
print(s1)
```

```python
print(s)

print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
print(s.dtype, s.name)

print(s.loc["A"])
print(s.loc["A":"C"])
print(s.iloc[0])
print(s.iloc[0:2])

print(s.at["A"])
print(s.iat[0])
```

```python
s["F"] = 6
print(s)

print(s["A"])
print(s[s < 3])
print(s.head())
print(s.tail())
print(s.head(2))
print(s.tail(1))
```

```python
import numpy as np

s = pd.Series(
    [10, 2, np.nan, None, 3, 4, 5],
    index=["A", "B", "C", "D", "E", "F", "G"],
    name="data",
)
print(s)
```

```python
print(s.head(3))
print(s.tail(6))
```

```python
print(s.describe())

print(s.mean())
print(s.sum())
print(s.std())
print(s.var())
print(s.min())
print(s.max())

print(s.median())
print(s.quantile(0.8))

s["H"] = 4
print(s.mode())
print(s.value_counts())
```
